```python
def churn_rate_table(data, column):

    result = (
        data.groupby(column)["churn_flag"]
        .agg(
            Customers="count",
            Churn_Rate="mean"
        )
        .sort_values(
            "Churn_Rate",
            ascending=False
        )
    )

    result["Churn_Rate"] *= 100

    return result.round(2)
```
